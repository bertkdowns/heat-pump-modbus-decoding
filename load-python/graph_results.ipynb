```python
# Run

import pandas as pd
from parse_csv import parse_csv

df = parse_csv('datalogs2.csv')
df_interpolated = df.interpolate(method='linear', limit_direction='both')

columns = {
"0_2013":	"Setpoint",
"0_2023":	"Water inlet Temp",
"0_2024":	"Water outlet Temp",
"0_2026":	"Ambient Temp",
"0_2027":	"Coil Temp",
"0_2031":	"Coil Temp",
"0_2043":	"Suction Temp",
"0_2044":	"Suction Temp",
"0_2047":	"Exhaust Air temp",
"0_2048":	"Exhaust Air temp",
"0_2059":	"Antifreeze temp",
"0_2063":	"Antifreeze temp",
"0_2087":	"Low pressure",
"0_2088":	"Low pressure",
"0_2111":	"Evap temp",
"0_2112":	"Superheat",
"0_2113":	"Evap temp",
"0_2114":	"Superheat",
}

# get just the columns we care about
df_selected = df_interpolated[list(columns.keys())]
df_selected = df_selected.rename(columns=columns)

df_selected
```

```python
# plot all temperatures
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
for column in df_selected.columns:
    if 'Temp' in column:
        plt.plot(df_selected.index, df_selected[column], label=column)

plt.legend()
plt.title('Temperature Readings Over Time')
plt.xlabel('Time')
```

```python
# Download as CSV
df_selected.to_csv('Processed Data.csv')
```
